#### Static Prompts

```python
from dotenv import load_dotenv
load_dotenv()
```

```python
from langchain_openai import ChatOpenAI
llm = ChatOpenAI(model="gpt-6-luna")
```

```python
prompts = [
    ("system", "You are an expert GenAI engineer. Be clear and concise."),
    ("user", "How do I build a GenAI chatbot? Which tech stack should I choose?")
]

res = llm.invoke(prompts)
print(res.content)
```
