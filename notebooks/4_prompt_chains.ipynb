#### Prompt Templates - Dynamic Prompts

```python
from langchain_openai import ChatOpenAI
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.output_parsers import StrOutputParser
```

```python
llm = ChatOpenAI(model="gpt-6-luna")
response_parser = StrOutputParser()
```

```python
prompts = ChatPromptTemplate.from_messages([
    {
        "role": "system",
        "content": "You are an expert {engineer} engineer."
    },
    {
        "role": "user",
        "content": "What is the difference between {query}?"
    }
])
```

```python
final_prompt = prompts.format_messages(engineer="Pythin", query="list and tuple")
# res = llm.invoke(final_prompt)
# print(res.content)
```

```python
res = llm.invoke(prompts.invoke({"engineer": "Python", "query": "list and tuple"}))
print(res.content)
```

```python
chains = prompts | llm | response_parser
res = chains.invoke({"engineer": "Python", "query": "dict and set"})
print(res)
```
