```python
from dotenv import load_dotenv

load_dotenv()
```

```python
from langchain.agents import create_agent
from langchain_community.utilities import GoogleSerperAPIWrapper
from langchain_core.tools import Tool
from langchain_groq import ChatGroq
from langgraph.checkpoint.memory import InMemorySaver
```

```python
model = ChatGroq(model="openai/gpt-oss-20b")
search = GoogleSerperAPIWrapper()

search_tool = Tool(
    name="web_search",
    description="Search the web for relevant information.",
    func=search.run
)

memory = InMemorySaver()
```

```python
agent = create_agent(
    model=model,
    tools=[search_tool],
    checkpointer=memory,
    system_prompt=(
        "You are a helpful assistant that can search the web "
        "for information. Return relevant information in a "
        "structured format. Remember the previous messages "
        "in the current conversation and use them when relevant."
    ),
)

config = {
    "configurable": {
        "thread_id": "user_1"
    }
}
```

```python
response = agent.invoke(
    {
        "messages": [
            {
                "role": "user",
                # "content": "Who is the Prime Minister of India?"
                "content": "Which political party does he belong to?"
            }
        ]
    },
    config=config
)

print("Assistant:", response["messages"][-1].content)

```
